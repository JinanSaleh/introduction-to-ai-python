```python
# Importing required libraries
import pandas as pd
import numpy as np

# Tools for splitting data
from sklearn.model_selection import train_test_split

# Tool for converting text into numeric features
from sklearn.feature_extraction.text import TfidfVectorizer

# Models
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression

# Evaluation metrics
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
# Load dataset
data = pd.read_csv("spam.csv", encoding="latin-1")

# Keeping only useful columns
data = data[['v1', 'v2']]

# Renaming columns for clarity
data.columns = ['label', 'message']

# Converting labels from text to numbers
# ham = 0, spam = 1
data['label'] = data['label'].map({'ham': 0, 'spam': 1})

# Showing first 5 rows
print(data.head())

# Show class distribution
print("\nClass distribution:")
print(data['label'].value_counts())
```

```python
# Spliting data into training and testing sets
# 80 percent for training, 20 percent for testing
X_train, X_test, y_train, y_test = train_test_split(
    data['message'],
    data['label'],
    test_size=0.2,
    random_state=42
)
```

```python
# Converting text into numerical features using TF-IDF
# Stoping words are removed to improve quality
vectorizer = TfidfVectorizer(stop_words='english')

# Fiting on training data and transform both sets
X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

# Printing shape of feature matrix
print("Training data shape:", X_train_tfidf.shape)
```

```python
#  Naïve Bayes model
nb_model = MultinomialNB()

# Training model on training data
nb_model.fit(X_train_tfidf, y_train)
```

```python
# Predicting on test data
nb_pred = nb_model.predict(X_test_tfidf)

# Printing results
print("Naïve Bayes Results")
print("Accuracy:", accuracy_score(y_test, nb_pred))

# Confusion matrix
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, nb_pred))

# Detailed classification report
print("\nClassification Report:")
print(classification_report(y_test, nb_pred))
```

```python
# Creating Logistic Regression model
lr_model = LogisticRegression(max_iter=1000)

# Training model
lr_model.fit(X_train_tfidf, y_train)
```

```python
# Predicting on test data
lr_pred = lr_model.predict(X_test_tfidf)

# Print results
print("Logistic Regression Results")
print("Accuracy:", accuracy_score(y_test, lr_pred))

# Confusion matrix
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, lr_pred))

# Detailed report
print("\nClassification Report:")
print(classification_report(y_test, lr_pred))
```

```python
# Predicting on test data
lr_pred = lr_model.predict(X_test_tfidf)

# Printing results
print("Logistic Regression Results")
print("Accuracy:", accuracy_score(y_test, lr_pred))

# Confusion matrix
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, lr_pred))

# Detailed report
print("\nClassification Report:")
print(classification_report(y_test, lr_pred))
```

```python
# Creating clean confusion matrix table for report
cm = confusion_matrix(y_test, nb_pred)

print("\nConfusion Matrix Table")
print("                Predicted Ham    Predicted Spam")
print(f"Actual Ham         {cm[0][0]}               {cm[0][1]}")
print(f"Actual Spam        {cm[1][0]}               {cm[1][1]}")
```

```python
# Geting feature names (words)
feature_names = vectorizer.get_feature_names_out()

# Geting log probabilities of words for spam class (class = 1)
spam_log_probs = nb_model.feature_log_prob_[1]

# Geting top 20 words with highest probability in spam
top_indices = spam_log_probs.argsort()[-20:][::-1]

print("Top Words Associated with Spam:\n")

for i in top_indices:
    print(feature_names[i])
```

```python
# Ranking Top Spam Words
print("Top Spam Words with Ranking:\n")

for rank, i in enumerate(top_indices, 1):
    print(f"{rank}. {feature_names[i]}")
```
